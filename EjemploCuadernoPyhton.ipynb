```python
#En las excepciones no capturadas por nosotros, el programa  lanza la exception (no nuestra excepción) y se detiene en el finally 
def indexador(objeto, indice):
    return objeto[indice]
try:
    indexador('Python', 10)
    print('Est no se ejecuta')
finally:
    print('Esto se ejecuta includo cuando salta la excepción')
print('Este print tampoco se ejecuta')
```

```python
def rev_sentencia(sentence):

    words = sentence.split(' ')
    reverse_sentence = ' '.join(reversed(words))

    return reverse_sentence

#Podemos hacer un print para llamar a la funión
#print(rev_sentencia("Hola mundo desde Python")) o lo siguiente enun fichero .py Python establece __name__ como
# "__main__" cuando el archivo se ejecuta directamente, pero si se importa como módulo, __name__ toma el nombre del módulo.

if __name__ == "--main__":
    input = "Hola mundo desde Python"
    print(rev_sentencia(input))
```

```python
#De esta manera podemos hacer copias independientes de una lista --> listas por comprensión.
la = [1, 2, 3]
lb = la[:]
lc = list(la)
ld = la.copy()

la = 'z'

print(la)
print(lb)
print(lc)
print(ld)
```

```python
#Los diccionartios también son mutables (como las listas) y se pueden copiar de forma independiente con el método copy()
thisdict ={
    "brand": "Seat Ibiza",
    "model": "Ibiza",
    "year": 2020
}
mydict = thisdict.copy()
print(mydict)
print(thisdict)

print(id(mydict))
print(id(thisdict))

mydict2 = thisdict
print(mydict2)
print(thisdict)
print(id(mydict2))
```
